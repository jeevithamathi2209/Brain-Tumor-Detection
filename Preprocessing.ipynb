```python
import pandas as pd 
df = pd.read_csv("Brain_Tumor_Prediction_Dataset.csv")
columns = [
    'Headache',
    'Nausea',
    'Vomiting',
    'Seizures',
    'Vision_Problems',
    'Memory_Loss',
    'Speech_Difficulty',
    'Balance_Problems',
    'Weakness',
    'Fatigue',
    'Brain_Tumor'
]
for col in columns:
    df[col] = df[col].replace({1 : 'Yes',
                 0 : 'No',
                 'YES' : 'Yes',
                 'NO' : 'No',
                 '1' : 'Yes',
                 '0' : 'No'  })
df.to_csv("brain_tumor_yes_no_dataset.csv", index=False)
print(df.head())
```

```python
df[columns] = df[columns].astype('string')
print(df.dtypes)
```

```python
df.isnull().sum()
```

```python
for col in columns:
    yes_count = (df[col] == 'Yes').sum()
    no_count = (df[col] == 'No').sum()
    total = 10000
    yes_percentage = (yes_count / total) * 100 if total > 0 else 0
    no_percentage = (no_count / total) * 100 if total > 0 else 0
    print(f"\n{col}")
    print(f"Yes Count      : {yes_count}")
    print(f"No Count       : {no_count}")
    print(f"Yes Percentage : {yes_percentage:.2f}%")
    print(f"No Percentage  : {no_percentage:.2f}%")
```

```python

df.replace(r'^\s*$', pd.NA, regex=True, inplace=True)

print("Missing Values Before Cleaning:")
print(df.isnull().sum())

df = df.bfill()
df = df.ffill()

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
print(df.head())
```

```python
for  col in columns:
    yes_count = (df[col] == 'Yes').sum()
    no_count = (df[col] == 'No').sum()
    total = 10000
    yes_percentage = (yes_count / total) * 100 if total > 0 else 0
    no_percentage = (no_count / total) * 100 if total > 0 else 0
    print(f"\n{col}")
    print(f"Yes Count      : {yes_count}")
    print(f"No Count       : {no_count}")
    print(f"Yes Percentage : {yes_percentage:.2f}%")
    print(f"No Percentage  : {no_percentage:.2f}%")
```

### Which symptom is most common

```python
symptoms = [
    'Headache',
    'Nausea',
    'Vomiting',
    'Seizures',
    'Vision_Problems',
    'Memory_Loss',
    'Speech_Difficulty',
    'Balance_Problems',
    'Weakness',
    'Fatigue'
]

# Count Yes values for each symptom
yes_counts = []

for col in symptoms:
    count = (df[col] == 'Yes').sum()
    yes_counts.append(count)

# Find most common symptom
max_count = max(yes_counts)
most_common = symptoms[yes_counts.index(max_count)]

print("Most Common Symptom:", most_common)
print("Count:", max_count)

# Visualization
plt.figure(figsize=(10, 5))
plt.bar(symptoms, yes_counts)

plt.title("Frequency of Symptoms")
plt.xlabel("Symptoms")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)

# Show values on bars
for i, v in enumerate(yes_counts):
    plt.text(i, v, str(v), ha='center')

plt.show()
```

```python
symptoms = [
    'Headache',
    'Nausea',
    'Vomiting',
    'Seizures',
    'Vision_Problems',
    'Memory_Loss',
    'Speech_Difficulty',
    'Balance_Problems',
    'Weakness',
    'Fatigue'
]

# Count Yes symptoms for each patient
df['Yes_Symptom_Count'] = (df[symptoms] == 'Yes').sum(axis=1)

# Patients with >= 5 symptoms
five_or_more = (df['Yes_Symptom_Count'] >= 5).sum()

# Patients with < 5 symptoms
less_than_five = (df['Yes_Symptom_Count'] < 5).sum()

print("Patients with >= 5 symptoms:", five_or_more)
print("Patients with < 5 symptoms:", less_than_five)

plt.figure(figsize=(6,6))

plt.pie(
    [five_or_more, less_than_five],
    labels=['>= 5 Symptoms', '< 5 Symptoms'],
    autopct='%1.1f%%'
)

plt.title("Patients with 5 or More Symptoms")
plt.show()
```

### Which symptom is most associated with Brain Tumor

```python
tumor_yes = df[df['Brain_Tumor'] == 'Yes']

symptom_counts = {}

for col in symptoms:
    symptom_counts[col] = (tumor_yes[col] == 'Yes').sum()

symptom_counts = dict(
    sorted(symptom_counts.items(),
           key=lambda x: x[1],
           reverse=True)
)
print(symptom_counts)
```

```python
import matplotlib.pyplot as plt

symptom_names = list(symptom_counts.keys())
counts = list(symptom_counts.values())

plt.figure(figsize=(10,6))
plt.barh(symptom_names, counts)

plt.title("Symptoms Most Associated with Brain Tumor")
plt.xlabel("Number of Brain Tumor Patients")
plt.ylabel("Symptoms")

plt.gca().invert_yaxis()  # Highest count on top

for i, v in enumerate(counts):
    plt.text(v, i, str(v))

plt.show()
```

### Top 3 symptoms among Brain Tumor patients

```python
tumor_yes = df[df['Brain_Tumor'] == 'Yes']

symptom_counts = {}

for symptom in symptoms:
    symptom_counts[symptom] = (tumor_yes[symptom] == 'Yes').sum()

top3_df = pd.DataFrame(
    symptom_counts.items(),
    columns=['Symptom', 'Count']
).sort_values(
    by='Count',
    ascending=False
).head(3)

print("Top 3 Symptoms Among Brain Tumor Patients")
print(top3_df)

# Percentage among tumor patients
top3_df['Percentage'] = (
    top3_df['Count'] / 10000
) * 100

print("\nWith Percentage")
print(top3_df)
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    top3_df['Symptom'],
    top3_df['Percentage']
)

plt.title("Top 3 Symptoms Among Brain Tumor Patients (%)")
plt.xlabel("Symptoms")
plt.ylabel("Percentage (%)")

for i, value in enumerate(top3_df['Percentage']):
    plt.text(i, value, f"{value:.1f}%", ha='center')

plt.show()
```

#### Which symptom is least common?

```python
symptom_counts = {}

for col in symptoms:
    symptom_counts[col] = (df[col] == 'Yes').sum()

least_common = min(symptom_counts, key=symptom_counts.get)

print("Least Common Symptom:", least_common)
print("Count:", symptom_counts[least_common])

# Show all counts
print("\nAll Symptom Counts:")
for symptom, count in symptom_counts.items():
    print(f"{symptom}: {count}")
```

### Which symptom combination appears most frequently?

```python
combination = df[symptoms].astype(str).agg('-'.join, axis=1)

# Count combinations
combination_counts = combination.value_counts()

# Most frequent combination
print("Most Frequent Symptom Combination:")
print(combination_counts.head(1))

# Top 10 combinations
print("\nTop 10 Symptom Combinations:")
print(combination_counts.head(10))
```

```python
import matplotlib.pyplot as plt

# Top 10 combinations
top10 = combination.value_counts().head(10)

plt.figure(figsize=(12,6))

plt.barh(
    range(len(top10)),
    top10.values
)

plt.yticks(
    range(len(top10)),
    [f'Combination {i+1}' for i in range(len(top10))]
)

plt.xlabel("Number of Patients")
plt.ylabel("Symptom Combination")
plt.title("Top 10 Most Frequent Symptom Combinations")

for i, v in enumerate(top10.values):
    plt.text(v, i, str(v))

plt.show()
```
